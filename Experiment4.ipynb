```python
import math
import numpy as np
import pandas as pd

data = [12, 15, 18, 20, 22, 25, 28, 30, 35, 40]

print("--- 1. Using NumPy ---")
mean_np = np.mean(data)
median_np = np.median(data)
std_population = np.std(data)          # ddof=0 (population standard deviation)
std_sample = np.std(data, ddof=1)      # ddof=1 (sample standard deviation)
var_sample = np.var(data, ddof=1)      # sample variance
data_range = np.ptp(data)              # peak-to-peak (max - min)
q25, q75 = np.percentile(data, [25, 75])
iqr = q75 - q25                        # Interquartile Range

print(f"Mean:                       {mean_np:.2f}")
print(f"Median:                     {median_np:.2f}")
print(f"Sample Std Dev (s):         {std_sample:.2f}")
print(f"Population Std Dev (sigma): {std_population:.2f}")
print(f"Sample Variance (s^2):      {var_sample:.2f}")
print(f"Range (Max - Min):          {data_range}")
print(f"IQR (Q3 - Q1):              {iqr:.2f}")


print("\n--- 2. Using Pandas (Full Summary) ---")
series = pd.Series(data)

summary = series.describe()
print(summary)


print("\n--- 3. Differences / Residuals (xi - mean) ---")

deviations = [x - mean_np for x in data]
for original, diff in zip(data, deviations):
    print(f"Value: {original:2d} | Deviation from mean: {diff:+6.2f}")
```
